# 04 - Scoring, statistics, failure analysis

```python
import pandas as pd
from trustbench.config import results_paths
from trustbench.scoring import score_results, compute_metrics_summary

MODE = "real"   # use "demo" only to look at simulator output
scored = score_results(MODE)
if MODE == "demo":
    print("SIMULATED DATA - do not quote these numbers")
compute_metrics_summary(scored)
```

```python
from trustbench.stats import (compute_confidence_intervals, compute_consistency, compute_cost_metrics,
                             all_paired_differences, generate_benchmark_charts)

ci = compute_confidence_intervals(scored)
print(ci)
generate_benchmark_charts(ci, results_paths(MODE)["charts"], demo=(MODE == "demo"))
print(compute_consistency(scored))
print(compute_cost_metrics(scored))
```

```python
# If clear_winner is False, say "no clear winner". Do not claim differences that may be chance.
all_paired_differences(scored)
```
